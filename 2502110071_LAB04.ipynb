IT10

```python
import random

accept = 0
rework = 0
scrap = 0

print("=== SELF-SORTING BENCH ===")

for i in range(100):

    diameter = random.uniform(24.80, 25.20)
    diameter = round(diameter, 2)

    if 24.95 <= diameter <= 25.05:
        print(diameter, "mm - ACCEPT")
        accept += 1

    elif 24.90 <= diameter < 24.95 or 25.05 < diameter <= 25.10:
        print(diameter, "mm - REWORK")
        rework += 1

    else:
        print(diameter, "mm - SCRAP")
        scrap += 1

print("\n=== SHIFT REPORT ===")

print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)

print("Total parts:", accept + rework + scrap)
```

```python
import random

accept = 0
rework = 0
scrap = 0
inspected = 0

mode = input("Choose mode (halt/skip): ").lower()

print("\n=== PRODUCTION STARTED ===")

for i in range(100):

    diameter = round(random.uniform(24.80, 25.20), 2)
    inspected += 1

    if 24.95 <= diameter <= 25.05:
        print(diameter, "mm - ACCEPT")
        accept += 1

    elif 24.90 <= diameter < 24.95 or 25.05 < diameter <= 25.10:
        print(diameter, "mm - REWORK")
        rework += 1

    else:
        print(diameter, "mm - SCRAP")
        scrap += 1

        if mode == "halt":
            print("SCRAP detected! Stopping machine.")
            break

        elif mode == "skip":
            print("SCRAP recorded. Continuing production.")
            continue

        else:
            print("Invalid mode selected.")
            break

print("\n=== SHIFT REPORT ===")

print("Parts inspected:", inspected)
print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)
print("Production stopped or completed.")
```

```python
import random

consecutive = 0
attempts = 0
max_attempts = 500

accept = 0
rework = 0
scrap = 0

print("=== RUN UNTIL GOOD ===")

while consecutive < 5 and attempts < max_attempts:

    diameter = round(random.uniform(24.80, 25.20), 2)
    attempts += 1

    if 24.95 <= diameter <= 25.05:
        print(diameter, "mm - ACCEPT")

        accept += 1
        consecutive += 1

    elif 24.90 <= diameter < 24.95 or 25.05 < diameter <= 25.10:
        print(diameter, "mm - REWORK")

        rework += 1
        consecutive = 0

    else:
        print(diameter, "mm - SCRAP")

        scrap += 1
        consecutive = 0

print("\n=== FINAL PRODUCTION REPORT ===")

print("Total attempts:", attempts)
print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)

if consecutive == 5:
    print("SUCCESS: Five consecutive ACCEPT parts achieved!")

else:
    print("SAFETY STOP: Maximum attempts reached.")
```
