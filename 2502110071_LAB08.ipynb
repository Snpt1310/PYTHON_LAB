```python
class Machine:
    def __init__(self, name, speed, limit):
        self.name = name
        self.speed = speed
        self.limit = limit
        self.__parts = 0

    def __str__(self):
        return self.name

    def run(self, quantity):
        if quantity < 0 or quantity > self.limit:
            print("Invalid quantity")
            return False

        self.__parts += quantity
        print(self.name, "processed", quantity, "parts")
        return True

    def status(self):
        return {
            "name": self.name,
            "speed": self.speed,
            "parts": self.__parts
        }


m1 = Machine("Machine 1", 100, 50)
m2 = Machine("Machine 2", 120, 60)

m1.run(20)
m2.run(30)

print(m1.status())
print(m2.status())
```

```python
machines = [
    Machine("Machine 1", 100, 50),
    Machine("Machine 2", 120, 60),
    Machine("Machine 3", 150, 70)
]

jobs = [10, 20, 15, 25, 30, 12, 18, 22, 14, 16]

for i, job in enumerate(jobs):
    machine = machines[i % 3]
    machine.run(job)

print("\nPLANT REPORT")

total = 0

for machine in machines:
    data = machine.status()
    print(data)
    total += data["parts"]

print("Plant total:", total)
```
