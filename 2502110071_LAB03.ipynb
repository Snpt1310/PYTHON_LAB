IT10

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    diameter = float(input("Enter diameter of part: "))

    if 9.5 <= diameter <= 10.5:
        print("ACCEPT")
        accept += 1

    elif 9.0 <= diameter < 9.5 or 10.5 < diameter <= 11.0:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

print("\n--- FINAL INSPECTION REPORT ---")
print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)
print("Total parts:", accept + rework + scrap)
```

```python
total = int(input("Enter total number of parts: "))
size = int(input("Enter box size: "))

boxes = total // size
loose = total % size

print("\n--- PALLET REPORT ---")
print("Full boxes:", boxes)
print("Loose parts:", loose)

print("Audit:", boxes * size + loose == total)
```

IT6

```python
accept = 0
rework = 0
scrap = 0

print("=== BATCH INSPECTION ===")

for i in range(10):
    diameter = float(input("Enter diameter of part: "))

    if 9.5 <= diameter <= 10.5:
        print("ACCEPT")
        accept += 1

    elif 9.0 <= diameter < 9.5 or 10.5 < diameter <= 11.0:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

# Pack only accepted parts

box_size = 4

boxes = accept // box_size
loose = accept % box_size

# Final shift report

print("\n=== END OF SHIFT REPORT ===")

print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)

print("\n=== PALLET REPORT ===")

print("Full boxes:", boxes)
print("Loose accepted parts:", loose)

print("Packing audit:", boxes * box_size + loose == accept)

print("\nTotal inspected:", accept + rework + scrap)
```
