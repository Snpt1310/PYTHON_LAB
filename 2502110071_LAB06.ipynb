```python
# Function 1: Inspect diameter
def inspect(dia):
    """Takes a diameter and returns its ACCEPT, REWORK or SCRAP label."""

    if 24.95 <= dia <= 25.05:
        return "ACCEPT"

    elif 24.90 <= dia < 24.95 or 25.05 < dia <= 25.10:
        return "REWORK"

    else:
        return "SCRAP"


# Function 2: Count inspection labels
def tally(labels):
    """Takes a list of labels and returns their counts in a dictionary."""

    counts = {
        "ACCEPT": 0,
        "REWORK": 0,
        "SCRAP": 0
    }

    for label in labels:
        counts[label] += 1

    return counts


# Function 3: Generate shift report
def report(shift, counts):
    """Takes a shift name and count dictionary and returns a report line."""

    return (
        f"{shift} SHIFT | "
        f"ACCEPT: {counts['ACCEPT']} | "
        f"REWORK: {counts['REWORK']} | "
        f"SCRAP: {counts['SCRAP']}"
    )


# Test Function 1
print("Test 1:", inspect(25.02))

# Test Function 2
labels = ["ACCEPT", "REWORK", "ACCEPT", "SCRAP"]
print("Test 2:", tally(labels))

# Test Function 3
counts = tally(labels)
print("Test 3:", report("DAY", counts))
```

```python
import random


# Function 1
def inspect(dia):
    """Takes a diameter and returns its inspection category."""

    if 24.95 <= dia <= 25.05:
        return "ACCEPT"

    elif 24.90 <= dia < 24.95 or 25.05 < dia <= 25.10:
        return "REWORK"

    else:
        return "SCRAP"


# Function 2
def tally(labels):
    """Takes inspection labels and returns category counts."""

    counts = {
        "ACCEPT": 0,
        "REWORK": 0,
        "SCRAP": 0
    }

    for label in labels:
        counts[label] += 1

    return counts


# Function 3
def report(shift, counts):
    """Takes a shift and counts and returns a formatted report."""

    return (
        f"{shift} SHIFT | "
        f"ACCEPT: {counts['ACCEPT']} | "
        f"REWORK: {counts['REWORK']} | "
        f"SCRAP: {counts['SCRAP']}"
    )


# Generate machine readings
# List: preserves all generated inspection results
labels = []

for i in range(10):

    dia = round(random.uniform(24.80, 25.20), 2)

    result = inspect(dia)

    labels.append(result)

    print("Part", i + 1, dia, result)


# Calculate production counts
counts = tally(labels)

# Display final report
print("\n=== SHIFT REPORT ===")
print(report("DAY", counts))
```

```python
# Function 1
def inspect(dia):
    """Takes a diameter and returns its inspection category."""

    if 24.95 <= dia <= 25.05:
        return "ACCEPT"

    elif 24.90 <= dia < 24.95 or 25.05 < dia <= 25.10:
        return "REWORK"

    else:
        return "SCRAP"


# Function 2
def lookup(catalogue, code):
    """Takes a catalogue and part code and returns diameter or a default."""

    return catalogue.get(code, "Part not found")


# Function 3
def scrap_codes(catalogue):
    """Takes a parts dictionary and returns a set of scrap part codes."""

    # Set: stores unique part codes that fail inspection
    return {
        code
        for code, dia in catalogue.items()
        if inspect(dia) == "SCRAP"
    }


# Dictionary: maps each unique part code to its diameter
catalogue = {
    "P101": 25.02,
    "P102": 24.93,
    "P103": 25.15,
    "P104": 24.98,
    "P105": 25.12,
    "P106": 25.01,
    "P107": 24.85
}


# Look up part information
print("P101 diameter:", lookup(catalogue, "P101"))

print("P999 diameter:", lookup(catalogue, "P999"))

# Find scrap part codes
print("Scrap part codes:", sorted(scrap_codes(catalogue)))
```

```python
import shoptools

logs = ["P101,25.02", "P102,24.93", "P103,25.15", "P104,24.98", "P105,25.12"]

catalogue = {log.split(",")[0]: float(log.split(",")[1]) for log in logs}

labels = [shoptools.inspect(dia) for dia in catalogue.values()]

counts = shoptools.tally(labels)

print(shoptools.report("DAY", counts))

print("Scrap codes:", sorted(shoptools.scrap_codes(catalogue)))
```
