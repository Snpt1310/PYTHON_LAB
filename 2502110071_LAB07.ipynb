```python
# Define Machine class
class Machine:

    def __init__(self, machine_id, name, limit):
        self.machine_id = machine_id
        self.name = name
        self.limit = limit
        self.parts = 0

    # Display machine nameplate
    def __str__(self):
        return (
            f"Machine ID: {self.machine_id}, "
            f"Name: {self.name}, "
            f"Limit: {self.limit} mm"
        )

    # Working method
    def inspect(self, dia):
        self.parts += 1

        if 24.95 <= dia <= 25.05:
            return "ACCEPT"
        elif 24.90 <= dia <= 25.10:
            return "REWORK"
        else:
            return "SCRAP"


# Commission two machines
machine1 = Machine("M01", "Lathe Machine", 25.05)
machine2 = Machine("M02", "Milling Machine", 25.05)

# Print machine nameplates
print(machine1)
print(machine2)

# Process parts
print("\nMachine 1:", machine1.inspect(25.02))
print("Machine 2:", machine2.inspect(24.93))

# Prove independent states
print("\nMachine 1 parts:", machine1.parts)
print("Machine 2 parts:", machine2.parts)
```

```python
class Machine:

    def __init__(self, machine_id, name, min_limit, max_limit):

        self.machine_id = machine_id
        self.name = name

        # Private machine settings
        self.__min_limit = min_limit
        self.__max_limit = max_limit

        # Private production counter
        self.__parts = 0
        self.__accept = 0
        self.__rework = 0
        self.__scrap = 0

    # Machine operation with validation
    def operate(self, dia):

        if not isinstance(dia, (int, float)) or isinstance(dia, bool):
            print("ERROR: Diameter must be numeric.")
            return "INVALID"

        if not self.__min_limit <= dia <= self.__max_limit:
            print("ERROR: Diameter outside machine operating range.")
            return "INVALID"

        self.__parts += 1

        if 24.95 <= dia <= 25.05:
            self.__accept += 1
            return "ACCEPT"

        elif dia < 24.95 or dia > 25.05:
            self.__rework += 1
            return "REWORK"

        else:
            self.__scrap += 1
            return "SCRAP"

    # Getter method
    def status(self):
        """Returns the current machine production status."""

        return {
            "Machine": self.name,
            "Total": self.__parts,
            "Accept": self.__accept,
            "Rework": self.__rework,
            "Scrap": self.__scrap
        }


# Create machine
machine = Machine("M01", "CNC Lathe", 24.90, 25.10)

# Test valid commands
print(machine.operate(25.02))
print(machine.operate(24.93))
print(machine.operate(25.08))

# Test invalid command
print(machine.operate(25.50))

# Display machine status
print("\n=== MACHINE STATUS ===")
print(machine.status())
```

```python
import random


class Machine:

    def __init__(self, machine_id, name):

        self.machine_id = machine_id
        self.name = name

        # Private counters
        self.__parts = 0
        self.__accept = 0
        self.__rework = 0
        self.__scrap = 0

    def operate(self, dia):
        """Processes one diameter and returns its inspection result."""

        if not isinstance(dia, (int, float)) or isinstance(dia, bool):
            return "INVALID"

        self.__parts += 1

        if 24.95 <= dia <= 25.05:
            self.__accept += 1
            return "ACCEPT"

        elif 24.90 <= dia <= 25.10:
            self.__rework += 1
            return "REWORK"

        else:
            self.__scrap += 1
            return "SCRAP"

    def status(self):
        """Returns machine production counters as a dictionary."""

        return {
            "Machine": self.name,
            "Total": self.__parts,
            "Accept": self.__accept,
            "Rework": self.__rework,
            "Scrap": self.__scrap
        }


# Commission machine
machine = Machine("M01", "CNC Lathe")

print("=== BATCH PRODUCTION ===")

# Generate and process 10 jobs
for i in range(10):

    dia = round(random.uniform(24.80, 25.20), 2)

    result = machine.operate(dia)

    print("Job", i + 1, "| Diameter:", dia, "|", result)


# Generate shift report
print("\n=== SHIFT REPORT ===")

report = machine.status()

for key, value in report.items():
    print(key, ":", value)
```

```python
class Machine:

    def __init__(self, machine_id, name):

        self.machine_id = machine_id
        self.name = name

        # Private production counters
        self.__parts = 0
        self.__accept = 0
        self.__rework = 0
        self.__scrap = 0

    def operate(self, dia):
        """Processes a part and returns its inspection category."""

        if not isinstance(dia, (int, float)) or isinstance(dia, bool):
            return "INVALID"

        self.__parts += 1

        if 24.95 <= dia <= 25.05:
            self.__accept += 1
            return "ACCEPT"

        elif 24.90 <= dia <= 25.10:
            self.__rework += 1
            return "REWORK"

        else:
            self.__scrap += 1
            return "SCRAP"

    def status(self):
        """Returns the machine name and production counters."""

        return {
            "ID": self.machine_id,
            "Name": self.name,
            "Parts": self.__parts,
            "Accept": self.__accept,
            "Rework": self.__rework,
            "Scrap": self.__scrap
        }
```

```python
import random
import shoptools

# List: stores three independent machine objects
machines = [
    shoptools.Machine("M01", "CNC Lathe"),
    shoptools.Machine("M02", "Milling Machine"),
    shoptools.Machine("M03", "Drilling Machine")
]

# Generate 10 production jobs
# List: preserves the order of production jobs
jobs = [
    round(random.uniform(24.80, 25.20), 2)
    for i in range(10)
]

# Allocate jobs to machines in rotation
for i, dia in enumerate(jobs):
    machine = machines[i % len(machines)]
    result = machine.operate(dia)

    print(machine.name, "|", dia, "|", result)

# Collect machine reports
# List: stores the status dictionary of every machine
reports = [machine.status() for machine in machines]

print("\n=== PLANT REPORT ===")

for report in reports:
    print(report["Name"], ":", report["Parts"], "parts")

# Calculate plant total
plant_total = sum(report["Parts"] for report in reports)

print("Plant total:", plant_total)

# Find busiest machine
busiest = max(machines, key=lambda m: m.status()["Parts"])

print("Busiest machine:", busiest.name)
print("Busiest machine parts:", busiest.status()["Parts"])
```
