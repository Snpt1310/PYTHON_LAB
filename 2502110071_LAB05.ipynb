```python
# List because we need to store and process all 10 log lines in order
logs = [
    "P101, 25.04, Day",
    " P102 , 24.98 , Day ",
    "P103,25.10, Night",
    " P104, 25.02, Day",
    "P105 , 24.95 , Night",
    " P106,25.08,Night ",
    "P107, 25.01, Day",
    "P108,24.99,Night",
    " P109 , 25.06 , Day ",
    "P110, 25.00, Night"
]

# List because diameters can contain duplicate values and we need all observations
diameters = []

for line in logs:
    part_code, dia, shift = [x.strip() for x in line.split(",")]
    dia = float(dia)
    diameters.append(dia)

print("Diameters:", diameters)
print("Count:", len(diameters))
print("Average:", sum(diameters) / len(diameters))
print("Minimum:", min(diameters))
print("Maximum:", max(diameters))
```

IT8

```python
# List because we need to store individual diameters
logs = [
    "P101, 25.04, Day",
    "P102, 24.98, Day",
    "P103, 25.12, Night",
    "P104, 25.02, Day",
    "P105, 24.95, Night"
]

# Lists because each category stores individual diameter values
accepted = []
rework = []
scrap = []

# List because we need the codes of scrapped parts
scrapped_parts = []

for line in logs:
    part, dia, shift = [x.strip() for x in line.split(",")]
    dia = float(dia)

    if 25.00 <= dia <= 25.05:
        accepted.append(dia)
    elif 24.95 <= dia <= 25.10:
        rework.append(dia)
    else:
        scrap.append(dia)
        scrapped_parts.append(part)

print("Accepted count:", len(accepted))
print("Rework count:", len(rework))
print("Scrap count:", len(scrap))

print("Accepted average:", sum(accepted) / len(accepted))
print("Scrapped parts:", scrapped_parts)
```

```python
# Lists because these contain individual shift records
day = ["P101,25.04,Day", "P102,24.98,Day", "P103,25.12,Day"]
night = ["P104,25.02,Night", "P105,25.12,Night", "P106,24.99,Night"]

# Lists because we need to store scrap part codes from each shift
day_scrap = []
night_scrap = []

for line in day:
    part, dia, shift = line.split(",")
    dia = float(dia)

    if dia < 24.95 or dia > 25.10:
        day_scrap.append(part)

for line in night:
    part, dia, shift = line.split(",")
    dia = float(dia)

    if dia < 24.95 or dia > 25.10:
        night_scrap.append(part)

# Sets because we need unique part codes for comparison
day_set = set(day_scrap)
night_set = set(night_scrap)

print("Day scrap:", day_set)
print("Night scrap:", night_set)

print("Common scrap:", day_set & night_set)
print("Unique to one shift:", day_set ^ night_set)
```
